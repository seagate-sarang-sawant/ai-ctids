```python
import numpy as np
import pandas as pd
import seaborn as sns
import missingno as msno
sns.set(style='darkgrid')
import matplotlib.pyplot as plt
```

```python
Project_root = "/Users/sarsawant/workspace/ai-ctids"

data1 = pd.read_csv(f'{Project_root}/data/Monday-WorkingHours.pcap_ISCX.csv')
data2 = pd.read_csv(f'{Project_root}/data/Tuesday-WorkingHours.pcap_ISCX.csv')
data3 = pd.read_csv(f'{Project_root}/data/Wednesday-workingHours.pcap_ISCX.csv')
data4 = pd.read_csv(f'{Project_root}/data/Thursday-WorkingHours-Morning-WebAttacks.pcap_ISCX.csv')
data5 = pd.read_csv(f'{Project_root}/data/Thursday-WorkingHours-Afternoon-Infilteration.pcap_ISCX.csv')
data6 = pd.read_csv(f'{Project_root}/data/Friday-WorkingHours-Morning.pcap_ISCX.csv')
data7 = pd.read_csv(f'{Project_root}/data/Friday-WorkingHours-Afternoon-PortScan.pcap_ISCX.csv')
data8 = pd.read_csv(f'{Project_root}/data/Friday-WorkingHours-Afternoon-DDos.pcap_ISCX.csv')
```

```python
data_list = [data1, data2, data3, data4, data5, data6, data7, data8]

print('Data dimensions: ')
for i, data in enumerate(data_list, start = 1):
  rows, cols = data.shape
  print(f'Data{i} -> {rows} rows, {cols} columns')
```

```python
data = pd.concat(data_list)
rows, cols = data.shape

print('New dimension:')
print(f'Number of rows: {rows}')
print(f'Number of columns: {cols}')
print(f'Total cells: {rows * cols}')
```

```python
# Deleting dataframes after concating to save memory
for d in data_list: del d
```

```python
# Renaming the columns by removing leading/trailing whitespace
col_names = {col: col.strip() for col in data.columns}
data.rename(columns = col_names, inplace = True)
```

```python
dups = data[data.duplicated()]
print(f'Number of duplicates: {len(dups)}')
```

```python
data.drop_duplicates(inplace = True)
data.shape
```

```python
missing_val = data.isna().sum()
print(missing_val.loc[missing_val > 0])
```

```python
# Checking for infinity values
numeric_cols = data.select_dtypes(include = np.number).columns
inf_count = np.isinf(data[numeric_cols]).sum()
print(inf_count[inf_count > 0])
```

```python
# Replacing any infinite values (positive or negative) with NaN (not a number)
print(f'Initial missing values: {data.isna().sum().sum()}')

data.replace([np.inf, -np.inf], np.nan, inplace = True)

print(f'Missing values after processing infinite values: {data.isna().sum().sum()}')
```

```python
missing = data.isna().sum()
print(missing.loc[missing > 0])
```

```python
med_flow_bytes = data['Flow Bytes/s'].median()
med_flow_packets = data['Flow Packets/s'].median()

print('Median of Flow Bytes/s: ', med_flow_bytes)
print('Median of Flow Packets/s: ', med_flow_packets)
```

```python
# Filling missing values with median
data['Flow Bytes/s'].fillna(med_flow_bytes, inplace = True)
data['Flow Packets/s'].fillna(med_flow_packets, inplace = True)
     
```

```python
print('Number of \'Flow Bytes/s\' missing values:', data['Flow Bytes/s'].isna().sum())
print('Number of \'Flow Packets/s\' missing values:', data['Flow Packets/s'].isna().sum())
```
